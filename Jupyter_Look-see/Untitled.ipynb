```python
# Total number of passengers
total_pass = len(titanic)
total_pass = titanic.shape[0]


# Number of survivors (tips_m = tips[tips['sex'] == "Male"])
all_survivors = titanic[titanic['survived'] == 1]
total_survivors = len(all_survivors) 


# Number of fatalities (tips_m = tips[tips['sex'] == "Male"])
all_fatalities = titanic[titanic['survived'] == 0]
total_fatalities = len(all_fatalities)


# Find the proportion of survivors 
prop_survivors = total_survivors/total_pass
prop_survivors 
```
